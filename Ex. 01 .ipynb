```python
saved_img = cv2.imread('Eagle.png', 0)
color_img = cv2.cvtColor(saved_img, cv2.COLOR_GRAY2RGB)
```

10. Read in the image ('Apollo-11-launch.jpg').

```python
apollo = cv2.imread('Apollo-11-launch.jpg')
```

11. Add the following text to the dark area at the bottom of the image (centered on the image):

```python
text = 'Apollo 11 Saturn V Launch, July 16, 1969'
font_face = cv2.FONT_HERSHEY_PLAIN

h, w = apollo.shape[:2]
text_width = cv2.getTextSize(text, font_face, 1.5, 2)[0][0]
x = (w - text_width) // 2
y = h - 20

cv2.putText(apollo, text, (x, y), font_face, 1.5, (255, 255, 255), 2)
```

12. Draw a magenta rectangle that encompasses the launch tower and the rocket.

```python
rect_color = (255, 0, 255)

cv2.rectangle(apollo, (100, 50), (500, 500), rect_color, 3)
```

13. Display the final annotated image.

```python
plt.imshow(cv2.cvtColor(apollo, cv2.COLOR_BGR2RGB))
plt.title('Final Annotated Image')
plt.axis('off')
plt.show()
```

14. Read the image ('boy.jpg').

```python
img = cv2.imread('boy.jpg')
```

15. Adjust the brightness of the image.

```python
# Create a matrix of ones (with data type float64) 
matrix_ones = np.ones(img.shape, dtype=np.float64)
```

16. Create brighter and darker images.

```python
img_brighter = cv2.add(img, matrix_ones, dtype=cv2.CV_8U)
img_darker = cv2.subtract(img, matrix_ones, dtype=cv2.CV_8U)
```

17. Display the images (Original Image, Darker Image, Brighter Image).

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(img, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(img_darker, cmap='gray')
plt.title('Darker Image')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(img_brighter, cmap='gray')
plt.title('Brighter Image')
plt.axis('off')

plt.show()
```

18. Modify the image contrast.

```python
# Create two higher contrast images using the 'scale' option with factors of 1.1 and 1.2 (without overflow fix)
matrix1 = np.ones(img.shape, dtype=np.float64) * 1.1
matrix2 = np.ones(img.shape, dtype=np.float64) * 1.2

img_higher1 = cv2.multiply(img, matrix1, dtype=cv2.CV_8U)
img_higher2 = cv2.multiply(img, matrix2, dtype=cv2.CV_8U)
```

19. Display the images (Original, Lower Contrast, Higher Contrast).

```python
matrix_lower = np.ones(img.shape, dtype=np.float64) * 0.8
img_lower = cv2.multiply(img, matrix_lower, dtype=cv2.CV_8U)

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(img, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(img_lower, cmap='gray')
plt.title('Lower Contrast')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(img_higher2, cmap='gray')
plt.title('Higher Contrast')
plt.axis('off')

plt.show()
```

20. Split the image (boy.jpg) into the B,G,R components & Display the channels.

```python
img = cv2.imread('boy.jpg')

b, g, r = cv2.split(img)

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(b, cmap='gray')
plt.title('Blue Channel')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(g, cmap='gray')
plt.title('Green Channel')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(r, cmap='gray')
plt.title('Red Channel')
plt.axis('off')

plt.show()
```

21. _merged the R, G, B , display along with orginal image 

```python
merged = cv2.merge([b, g, r])

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(merged, cv2.COLOR_BGR2RGB))
plt.title('Merged Image')
plt.axis('off')

plt.show()
```

22. Split the image into the H, S, V components &  Display the channels.

```python
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
h, s, v = cv2.split(hsv)

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(h, cmap='gray')
plt.title('Hue')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(s, cmap='gray')
plt.title('Saturation')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(v, cmap='gray')
plt.title('Value')
plt.axis('off')

plt.show()
```

23. _merged the H, S, V, display along with orginal image 

```python
merged_hsv = cv2.merge([h, s, v])
merged_img = cv2.cvtColor(merged_hsv, cv2.COLOR_HSV2RGB)

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(merged_img)
plt.title('Merged HSV Image')
plt.axis('off')

plt.show()
```
